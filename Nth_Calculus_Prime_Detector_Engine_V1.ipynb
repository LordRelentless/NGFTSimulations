```python
import time

def n_selector(k):
    return f"n_|{k}|"

def is_nth_prime_by_divisibility(k):
    if k < 2:
        return False
    for d in range(2, int(k ** 0.5) + 1):
        if k % d == 0:
            return False
    return True

def nth_prime_engine_range(start, end):
    structural_primes = []
    start_time = time.time()

    for k in range(start, end + 1):
        if is_nth_prime_by_divisibility(k):
            structural_primes.append(k)

    end_time = time.time()
    elapsed = end_time - start_time
    return structural_primes, elapsed

# 📊 Define your desired range (adjust these as needed)
start_range = 19_999_000_000
end_range = 20_000_000_000  # Try a small slice first for benchmarking

# 🚀 Run engine and capture timing
nth_primes, runtime = nth_prime_engine_range(start_range, end_range)

# 🧮 Output results
print(f"\nSelector-Driven Nth Primes from {start_range:,} to {end_range:,}:")
print(f"Total found: {len(nth_primes)}")
print(f"Runtime: {runtime:.2f} seconds")
print(f"First few primes in range: {nth_primes[:10]}")
```
